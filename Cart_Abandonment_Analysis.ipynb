```python
# Checking categorical values

categorical_cols = [
    'device_type',
    'payment_method',
    'product_category',
    'returning_customer',
    'coupon_used'
]

for col in categorical_cols:
    print(f"\n{col}:")
    print(df[col].value_counts())
```

```python
# Detecting outliers using IQR method

numerical_cols = df.select_dtypes(include='number').columns.drop('cart_abandoned')

outlier_summary = []

for col in numerical_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    ].shape[0]

    outlier_summary.append({
        'Column': col,
        'Lower Bound': round(lower_bound, 2),
        'Upper Bound': round(upper_bound, 2),
        'Outliers': outliers
    })

outlier_df = pd.DataFrame(outlier_summary)

display(outlier_df)
```

```python
# Visualizing outliers using piechart

plt.figure(figsize=(15, 8))

for i, col in enumerate(categorical_cols, 1):
    plt.subplot(3, 3, i)

    df[col].value_counts().plot(
        kind='pie',
        autopct='%1.1f%%',
        startangle=90
    )

    plt.title(col)
    plt.ylabel('')

plt.tight_layout()
plt.show()
```

```python
# Checking binary columns

binary_cols = [
    'returning_customer',
    'coupon_used',
    'cart_abandoned'
]

for col in binary_cols:
    print(f"\n{col}:")
    print(df[col].value_counts())
```

```python
# Checking extra spaces in categorical columns

for col in df.select_dtypes(include='object').columns:
    spaces = df[col].astype(str).str.strip().ne(df[col].astype(str)).sum()
    print(f"{col}: {spaces} values with extra spaces")
```

```python
# Checking binary columns

binary_cols = [
    'returning_customer',
    'coupon_used',
    'cart_abandoned'
]

for col in binary_cols:
    print(f"\n{col}:")
    print(df[col].value_counts())
```

```python
# Final dataset validation

print("Dataset Shape:", df.shape)

print("Missing Values:", df.isnull().sum().sum())

print("Duplicate Rows:", df.duplicated().sum())

print("Unique Session IDs:", df['session_id'].nunique())
```

```python
# Device type distribution

plt.figure(figsize=(8, 5))

sns.countplot(data=df, x='device_type')

plt.title('Distribution of Device Types')
plt.xlabel('Device Type')
plt.ylabel('Number of Sessions')

plt.show()
```

```python
# Payment method distribution

plt.figure(figsize=(9, 5))

sns.countplot(
    data=df,
    x='payment_method',
    order=df['payment_method'].value_counts().index
)

plt.title('Distribution of Payment Methods')
plt.xlabel('Payment Method')
plt.ylabel('Number of Sessions')
plt.xticks(rotation=30)

plt.show()
```

```python
# Product category distribution

plt.figure(figsize=(9, 5))

sns.countplot(
    data=df,
    x='product_category',
    order=df['product_category'].value_counts().index
)

plt.title('Distribution of Product Categories')
plt.xlabel('Product Category')
plt.ylabel('Number of Sessions')
plt.xticks(rotation=30)

plt.show()
```
